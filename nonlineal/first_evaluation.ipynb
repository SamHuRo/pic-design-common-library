As is well known, the generation of nonlinear optical phenomena typically requires intense electric fields, because the nonlinear susceptibilities of materials are, in general, considerably smaller than the linear susceptibility. In particular, the second-order susceptibility, $\chi^{(2)}$, is typically on the order of $10^{-12}\,\text{m/V}$, whereas the third-order susceptibility, $\chi^{(3)}$, can be on the order of $10^{-24}\,\text{m}^2/\text{V}^2$. For this reason, strong electric fields are required to obtain significant nonlinear polarization. However, this does not guarantee that the generated wave will have high intensity at the material’s output, since its generation efficiency also depends on the interaction between the waves involved and the propagation conditions within the medium

During nonlinear processes, new frequencies are generated that can interfere constructively or destructively, depending on their phase relationships. To promote efficient conversion of optical energy, the phase-matching condition between the participating waves must be satisfied. This condition ensures that the phase difference between the generated waves and the nonlinear polarization that produces them remains constant—or varies sufficiently slowly—along the direction of propagation. In this way, the contribution of the nonlinear polarization builds up constructively, allowing the amplitude of the generated field to grow approximately linearly with the propagation distance and its intensity to increase quadratically, under ideal conditions and in the non-exhausted regime of the incident wave.

Consequently, phase matching enables efficient energy transfer from the incident waves to the generated wave, maximizing the efficiency of the nonlinear conversion process. In conclusion, it is not enough to have a strong incident electric field; it is also essential to satisfy the phase-matching condition to promote the efficient generation of new frequencies and obtain a strong optical output signal.

We have a mm2-type crystal, which is an orthorhombic $C_{2v}$ crystal with two mirror planes:
- \(yz\), corresponding to reflection $x\rightarrow -x$.
- \(xz\), corresponding to reflection $y\rightarrow -y$.

$$
\chi_{ijk} = \sum_{a \beta c} R_{ia} R_{j \beta} R_{kc} \chi_{a \beta c}

For every crystal symmetry R, we have that:

* $R$ is the matrix describing a reflection.

* $i,j,k$ are the three indices of the susceptibility tensor.

* The equation says that the tensor before and after the reflection must be identical.

Both mirrors are diagonal matrices, so each can be written as $R_{ia} = s_{i} \delta_{ia}$, with $s_i =  \pm 1$. Since $R$ is diagonal we have that the sum is going to be and the susceptibility tensor must remain unchanged when we apply a symmetry operation of the crystal:

$$
\chi'_{ijk} = R_{ii} R_{j j} R_{kk} \chi_{ijk}
$$

This tells us that each tensor index contributes a factor of either $+1$ or $-1$.

This means that for $M_X$ every $x$ is a $-1$, and $y,z$ is a $+1$, for example $\chi'_{xxz}$, applying the reflection rule we have that:

$$
\chi'_{xxz} = (-1)(-1)(+1) \chi_{xxz}
$$

$$
\chi'_{xxz} =  \chi_{xxz}
$$

For $\chi'_{xyz}$ we have:

$$
\chi'_{xyz} = (-1)(+1)(+1) \chi_{xyz}
$$

$$
\chi'_{xyz} = - \chi_{xyz}
$$

$$
2 \chi_{xyz} = 0
$$

$$
\chi_{xyz} = 0
$$

So:
* An even number of $x$'s gives $+1$.
* An odd number of $x$'s gives $-1$.

This also applies to $M_{\gamma}$.

For a component to be non zero we have that:

$$
R_{ii} R_{jj} R_{kk} = +1
$$

So the possible combinations are:

* If there are two $x$'s, the third index must be $z$, because we cannot have an odd number of $y$'s.
* If there are two $y$'s, the third index must be $z$.
* If there are zero $x$'s and zero $y$'s, all three indices must be $z$.

Applying both conditions to the 27 Cartesian components gives seven potentially nonzero components:

$$\chi_{zzz}^{(2)};
\chi_{xxz}^{(2)};
\chi_{xzx}^{(2)};
\chi_{zxx}^{(2)};
\chi_{yyz}^{(2)};
\chi_{yzy}^{(2)};
\chi_{zyy}^{(2)}$$

Consequently, the (mm2) crystal symmetry reduces the number of potentially nonzero Cartesian susceptibility components from 27 to 7, without imposing any additional permutation or Kleinman symmetry.

For a fixed $\theta= 40°$ we have:

$$
d_{eff} = - \sin (40°) (2 \sin^{2} \phi + 4 \cos^{2} \phi) = - \sin (40°) (2 (1 - \cos^{2}) \phi + 4 \cos^{2} \phi) = - \sin (40°) (2 + 2 \cos^{2} \phi) 
$$

We know that $ 0 \leq \cos^{2} \phi \leq 1$, to find the maximum $|d_{eff}|$ that $\cos^{2} \phi = 1$, this happens when $\phi = 0° \text{or} 180°$ and to find the minimum, it is when  $\cos^{2} \phi = 0$, this happens when $\phi = 90° \text{or} 270°$, so:

$$
|d_{eff}|_{max} = \sin (40°) (2 + 2) = 4 \sin(40°) \approx 2.571

$$

$$
|d_{eff}|_{min} = \sin (40°) (2) = 2 \sin(40°) \approx 1,286
$$

The negative sign indicates the phase or sign of the projected nonlinear polarization relative to the chosen polarization-vector conventions. Reversing the sign of the effective coefficient corresponds to a phase change of $\pi$ in the nonlinear source, all else being equal.

For a crystal with a symmetry of 4 mm, we have that the matrix:

$$
d = 
\begin{bmatrix}

So $d \vec{v}$:

$$
d \vec{v} = A_o^{2} \begin{bmatrix}

0 & 0 & 0 & d_{15} & 0 & 0 \\
d_{31} & d_{31} & d_{33} & 0 & 0 & 0 \\
\end{bmatrix} 

\begin{pmatrix}

\end{pmatrix}

We can see that with $4 mm $symmetry, the dependence on $\phi$ disappears; therefore, given a $\theta$ the effective nonlinearity is independent of $\phi$, because the fourfold rotational symmetry makes the $x$ - and $y$-directions equivalent.

A nonzero effective nonlinearity means that a nonlinear polarization source can drive the second-harmonic field, but efficient generation also requires the generated wave to remain in phase with that source as it propagates through the crystal.

For second-harmonic generation, the phase-matching condition is
$$\Delta k=k_{2\omega}-2k_\omega=0$$


Since $k=2\pi n/\lambda$, with $\lambda\$) the vacuum wavelength, this can be written as

$$\Delta k=
\frac{2\pi n_{2\omega}}{\lambda_{2\omega}}
-\frac{4\pi n_\omega}{\lambda_\omega}$$


Because $\lambda_{2\omega}=\lambda_\omega/2$, this simplifies to

$$n_{2\omega}=n_\omega$$

for collinear propagation, where the indices correspond to the appropriate polarization modes at their respective frequencies.

## d.

One limitation of the supplied polarization basis is that it is an idealized approximation and is not necessarily valid for a general anisotropic dielectric. The wave normal $\vec{k}$ describes the direction of phase propagation, while the electric field $\vec{E}$ and displacement field $\vec{D}$ are generally not parallel because $\vec{D} =\varepsilon_0\boldsymbol{\varepsilon}_r \vec{E}$. Maxwell's equations require $\vec{k} \cdot \vec{D}=0$, but $\vec{E}$ is not necessarily perpendicular to $\vec{k}$. Furthermore, the energy flow, given by the Poynting vector, need not be parallel to the wave normal, producing spatial walk-off. Therefore, in a general anisotropic dielectric, the actual polarization eigenvectors and energy-flow direction must be determined from the dielectric tensor and Maxwell's equations.

## b.

```python
import numpy as np

EPSILON = 8.8541878128e-12 # F/m
C = 299792458 # m/s


def I_2(
    wavelength,
    d_eff,
    n1,
    n2,
    I1,
    L,
    delta_k
):
    """
    Definition of function:
        Calculates the output intensity I2 based on the input intensity,
        wavelength, effective nonlinear coefficient, refractive indices,
        interaction length, and phase mismatch.

    Args:
        wavelength : Wavelength of the input light (m).
        d_eff: Effective nonlinear coefficient.
        n1: Refractive index of the input wave.
        n2: Refractive index of the output wave.
        I1: Input intensity (W/m^2).
        L: Interaction length (m).
        delta_k: Phase mismatch (1/m).

    Returns:
        I2: Calculated output intensity (W/m^2).
    """
    omega = 2 * np.pi * C / wavelength 

    arg1 = (2 * omega**2 * d_eff**2) / (n1**2 + n2 * EPSILON * C**3)
    arg2 = I1**2 * L**2
    x = delta_k * L / 2
    arg3 = np.sinc(x / np.pi)**2

    I2 = arg1 * arg2 * arg3

    return I2

def etta(
        I1,
        I2,
):
    """
    Definition of function:
        Calculates the conversion efficiency (etta) as the ratio
        of the output intensity to the input intensity.

    Args:
        I1: Input intensity (W/m^2).
        I2: Output intensity (W/m^2).

    Returns:
        Etta: Conversion efficiency (dimensionless).
    """
    Etta = I2 / I1
    return Etta
```

### For $W/m^{2}$

```python
n_1 = 1.65
n_2 = 1.67
d_eff = 2 # pm/V
L = 2 # mm

lambda_1 = 1064 # nm
I1 = 1 #MW/cm^2
delta_k = [0, 600]# 1/m

I2_list = []
etta_list = []

for i in delta_k:
    I2 = I_2(
        wavelength=lambda_1 * 1e-9,
        d_eff=d_eff * 1e-12,
        n1=n_1,
        n2=n_2,
        I1=I1 * 1e6 * 1e4,
        L=L * 1e-3,
        delta_k=i
    )

    I2_list.append(I2)
    print(f"Output intensity I2 for delta_k = {i} [1/m]: {I2:.4e} W/m^2")

    etta_value = etta(I1 * 1e6 * 1e4, I2)
    etta_list.append(etta_value)
    print(f"Conversion efficiency (etta): {etta_value:.4e}")
```

### $L_{coherence} = \dfrac{\pi}{|\Delta k|}$

```python
def L_coh(delta_k):
    """
    Definition of function:
        Calculates the coherence length based on the phase mismatch.

    Args:
        delta_k: Phase mismatch (1/m).

    Returns:
        L_coh: Coherence length (m).
    """
    L_coh = np.pi / np.abs(delta_k)
    return L_coh
```

```python
print(f"Coherence length L_coh: {L_coh(600):.4e} m")
```

The coherence length is the distance over which the phase difference between the generated second-harmonic field and its driving nonlinear polarization changes by $\pi$. The reason for this behavior is that if $L$ is greater than approximately $1/\Delta k$, the output wave can get out of phase with its driving polarization,
and power can flow from the $\omega_3$ wave back into the $\omega_1$ and $\omega_2$ waves.

```python
def I2_delta_I2_0(
    k_values,
    crystal_length
):
    """
    Definition of function:
        Calculates the ratio of output intensity I2 to input intensity I2_0
        based on the phase mismatch and interaction length.

    Args:
        delta_k: Phase mismatch (1/m).
        L: Interaction length (m).
    Returns:
        ratio: Ratio of output intensity to input intensity (dimensionless).
    """

    x = k_values * crystal_length / 2
    I_norm = np.sinc(x / np.pi)**2

    return I_norm
```

At the original intensity, the predicted conversion efficiency was very small, so assuming that the fundamental pump remains constant was reasonable.

At the increased intensity, the predicted efficiency is approximately $9.25\%$. This is no longer negligible. A noticeable amount of energy may be transferred from the fundamental wave to the second harmonic, meaning the fundamental intensity can decrease as the waves propagate through the crystal.

The undepleted-pump assumption would then become less accurate. A more complete model should solve the coupled propagation equations for both the fundamental and second-harmonic fields, allowing the pump to deplete and the generated field to feed back into the interaction.

The predicted $9.25\%$ is therefore an extrapolation of the simplified model, not a guaranteed experimental result.

Here are two additional physical effects that an limit conversion in a real experiment:

1. Pump depletion

As the second harmonic is generated, energy is transferred from the fundamental wave into the second-harmonic wave. The fundamental intensity decreases along the crystal, so the nonlinear source term, which depends on $A_1^2$, also decreases.

Consequently, the second-harmonic intensity no longer follows the simple $I_2\propto I_1^2L^2$ scaling indefinitely. A coupled-wave model is needed to describe this energy transfer.

2. Heating and thermal effects

At high optical intensities, absorption of some of the optical energy can heat the crystal. Temperature changes can alter its refractive indices and therefore its phase-matching condition. If the phase mismatch increases, the generated second-harmonic field accumulates less coherently, reducing conversion efficiency. Thermal gradients can also make phase matching vary across the beam.

\end{pmatrix}

2 \dfrac{2 \pi}{\lambda_s} n_p(\alpha) = 2 \dfrac{2 \pi}{\lambda_s} n_o(\lambda_s) \cos \beta

\beta = \cos^{-1} \left(\dfrac{n_p(\alpha)}{n_o(\lambda_s)} \right)

-1 \leq \dfrac{n_p(\alpha)}{n_o(\lambda_s)} \leq 1

If $n_p(\alpha)=n_o(\lambda_s)$, then $\beta=0^\circ$, corresponding to collinear propagation rather than a cone with a finite opening angle.

If $n_p(\alpha)>n_o(\lambda_s)$, the ratio exceeds one, so no real, exactly phase-matched degenerate cone exists under the assumptions of this problem. If the ratio is between zero and one, a real cone angle exists.

The optic axis determines the effective extraordinary refractive index of the pump, $n_p(\alpha)$. However, once the pump direction and this index are fixed, the ordinary signal and idler can satisfy vector matching at any azimuthal orientation, provided they have equal and opposite transverse wavevector components.

Consequently, the allowed signal and idler wavevectors trace a cone around the pump direction. Each signal direction has a matched idler direction on the opposite side of the cone. This rotationally symmetric locus follows from the assumed ordinary-wave dispersion and vector matching; it does not imply uniform brightness around the cone.

For angles below this threshold, there is no real, exactly phase-matched degenerate cone in the specified forward geometry.

$$
n_p(\alpha) = \left[\dfrac{\cos^{2}\alpha}{n_o^{2}(\lambda_p)} + \dfrac{\sin^{2}\alpha}{n_e^{2}(\lambda_p)} \right]^{-1/2}
$$

```python
# Wavelengths in micrometers
no_p = n_o(0.405)
ne_p = n_e(0.405)
no_s = n_o(0.810)

def n_p(alpha_deg, no, ne):
    a = np.radians(alpha_deg)
    return 1 / np.sqrt(
        np.cos(a)**2 / no**2
        + np.sin(a)**2 / ne**2
    )

# Requested angles
for a in [28, 29, 29.5, 30, 31]:
    ratio = n_p(a, no_p, ne_p) / no_s
    if ratio > 1:
        result = "No matched degenerate cone"
    else:
        result = f"beta = {np.degrees(np.arccos(ratio)):.3f} deg"
    print(f"alpha = {a:4.1f} deg, "
          f"n_p = {n_p(a, no_p, ne_p):.6f}, {result}")

# Plot only real solutions
alpha = np.linspace(27, 32, 1000)
ratio = np.array([n_p(a, no_p, ne_p)/no_s for a in alpha])
valid = ratio <= 1
beta = np.full_like(alpha, np.nan)
beta[valid] = np.degrees(np.arccos(ratio[valid]))

plt.plot(alpha, beta)
plt.axvline(alpha0, linestyle="--",
            label=f"Threshold = {alpha0:.3f} deg")
plt.xlim(27, 32)
plt.xlabel("Optic-axis angle alpha (degrees)")
plt.ylabel("Internal cone angle beta (degrees)")
plt.title("Degenerate cone angle vs optic-axis angle")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

## e.

In a classical coupled-wave model, the pump, signal, and idler interact through the nonlinear susceptibility. Schematically, the signal and idler equations contain terms of the form

$$\frac{dA_s}{dz}\propto A_p A_i^*e^{i\Delta kz}$$

$$\frac{dA_i}{dz}\propto A_p A_s^*e^{i\Delta kz}$$


where $A_p,A_s,A_i$ are the complex field amplitudes and

$$\Delta k=k_p-k_s-k_i$$


is the longitudinal phase mismatch for collinear propagation (or the corresponding longitudinal mismatch for noncollinear waves).

Suppose the pump is present, but the signal and idler start at exactly zero:

$$A_s(0)=0,\qquad A_i(0)=0$$

In these classical equations, the signal derivative is proportional to $A_i^*$, and the idler derivative is proportional to $A_s^*$. Both derivatives are therefore zero when both fields are zero.

Consequently,

$$\frac{dA_s}{dz}\propto 0$$

$$\frac{dA_i}{dz}\propto 0$$

$$A_s(z)=A_i(z)=cte$$

We see that it will remain constant throughout the entire crystal it remains a solution throughout the crystal, even when the wavevectors satisfy phase matching. A deterministic classical model with no initial signal or idler field does not spontaneously generate those fields from exactly zero amplitudes. In practice, an initial seed or fluctuations are needed to initiate the process in such a model.

The calculations in parts (a)–(d) establish the classically allowed frequency and wavevector geometry.

In particular, they establish:
- Frequency conservation: $\omega_p=\omega_s+\omega_i$.
- Wavevector matching: $\mathbf{k}_p=\mathbf{k}_s+\mathbf{k}_i$.
- Allowed emission directions: the internal cone angles and, in part (c), the corresponding external angles.
- Nondegenerate geometry: the signal and idler can have different wavelengths and different cone half-angles while satisfying vector matching.

However, these calculations do not establish the spontaneous emission rate, the brightness of the cone, or the probability of producing a signal–idler pair. Those require a description of how the initially absent fields become populated, beyond the deterministic classical model with exactly zero initial signal and idler amplitudes.

For a finite crystal of length $L$, contributions to the generated field accumulate along the propagation direction. If the longitudinal phase mismatch is $\Delta k$, contributions from different positions acquire different phases.

For a constant nonlinear source, the relevant longitudinal integral is proportional to

$$\mathcal A(L)\propto
\int_0^L e^{i\Delta kz}\,dz$$

The corresponding intensity factor is proportional to

$$
I(L)\propto
L^2\operatorname{sinc}^2
\left(\frac{\Delta kL}{2}\right)
$$

At perfect longitudinal phase matching, $\Delta k=0$, the contributions add coherently, and the intensity factor reaches its maximum.

When $\Delta k\neq0$, the contributions gradually become out of phase. They partially cancel, reducing the output. Nevertheless, a finite crystal does not require mathematically exact longitudinal matching: appreciable output can occur over a range of mismatch values.

The first zeros of the intensity factor occur when

$$\frac{|\Delta k|L}{2}=\pi$$

so

$$|\Delta k|=\frac{2\pi}{L}$$

Thus, a longer crystal generally gives a narrower longitudinal phase-matching acceptance, while a shorter crystal allows a broader range of mismatches.

A finite crystal length replaces idealized exact longitudinal matching with a finite acceptance width. The generated intensity follows a sinc-squared dependence on longitudinal phase mismatch in the constant-source approximation, with a narrower acceptance for longer crystals.
